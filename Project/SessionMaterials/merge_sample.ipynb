```python
import pandas as pd
import numpy as np
```

* Define two simple DataFrames
    - df_orders: This acts as our primary (left) DataFrame, similar to df_settlement in the workshop.It contains the core records we want to enrich.   
    - df_customer_info: This acts as our secondary (right) DataFrame, similar to df_api or df_mongo.It provides additional details to be joined.

```python
df_orders = pd.DataFrame({
    'order_id': ['ORD001', 'ORD002', 'ORD003', 'ORD004', 'ORD005'],
    'customer_id': ['C101', 'C102', 'C101', 'C103', 'C105'], # C105 will not have matching info
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'location': ['NYC', 'LA', 'NYC', 'CHI', 'SF'] # This column name will conflict with df_customer_info
})
print("Orders DataFrame:")
display(df_orders)
```

```python
df_customer_info = pd.DataFrame({
    'customer_id': ['C101', 'C102', 'C103', 'C104'], # C104 will not have matching orders
    'customer_name': ['Alice', 'Bob', 'Charlie', 'David'],
    'email': ['alice@example.com', 'bob@example.com', 'charlie@example.com', 'david@example.com'],
    'location': ['New York', 'Los Angeles', 'Chicago', 'Houston'] # This column name will conflict with df_orders
})
print("Customer Info DataFrame:")
display(df_customer_info) 
```

* Perform a left merge
    - 'on=' specifies the common column(s) to join on.
    - 'how=' specifies the type of merge (e.g., 'left', 'right', 'inner', 'outer').
    - 'suffixes=' specifies what to append to conflicting column names from the left and right DataFrames, respectively.

```python
merged_df = pd.merge(df_orders, df_customer_info, on='customer_id', how='left', suffixes=('_order', '_cust'))
display(merged_df)
```
