```python
import pandas as pd
import numpy as np

data = {
    'tran_id_raw': ['A101', 'B202', 'A101', 'C303', 'B202', 'D404', 'A101'],
    'value': [100, 200, 110, 300, 210, 400, 120],
    'source': ['API', 'SET', 'MONGO', 'API', 'SET', 'API', 'SET']
   }

df_example = pd.DataFrame(data)
display(df_example)
```

```python
before_rows = len(df_example)
print(f"Number of rows before deduplication: {before_rows}")
```

```python
# Identify duplicate transaction IDs using .duplicated(keep='first')
# The .duplicated() method returns a Boolean Series:
#    - False: for the first occurrence of a unique 'tran_id_raw' (these are kept)
#    - True: for all subsequent occurrences of that 'tran_id_raw' (these are duplicates to be dropped)
#    - 'keep="first"' ensures that the first record encountered for a given ID is preserved.
dupe_mask = df_example['tran_id_raw'].duplicated(keep='first')
```

```python
print("Boolean Mask identifying duplicates (`True` means it's a duplicate to be dropped):")
print(dupe_mask)
```

```python
dropped_dupes = df_example[dupe_mask].copy()
```

```python
print("Rows identified as Duplicates (these will be dropped from the main DataFrame):")
print(dropped_dupes)
```

```python
df_example_cleaned = df_example[~dupe_mask].reset_index(drop=True)
df_example_cleaned
```

```python
after_rows = len(df_example_cleaned)
```

```python
print(f"Summary: Original rows: {before_rows}, Dropped duplicates: {len(dropped_dupes)}, Remaining rows: {after_rows}")
```
