```python
import pandas as pd
```

```python
# Let's test the PAN validation regex on a few examples
pan_samples = pd.Series([
'XXXXXXXXXXXX1234',  # Valid
'XXXXXXXXXXXXI234',  # Invalid (contains a letter 'I')
'OOOOOOOOOOOO1234',  # Invalid (doesn't start with 'X')
'XXXXXXXXXXXX123',   # Invalid (too short)
'XXXXXXXXXXXX12345'  # Invalid (too long)
])
display(pan_samples)
```

```python
# The regex pattern:
# ^      - Start of the string
# X{12}  - Exactly 12 'X' characters
# \d{4}  - Exactly 4 digit characters (0-9)
# $      - End of the string
pan_regex = r'^X{12}\d{4}$'

# Create a DataFrame to show the results clearly
pan_validation_results = pd.DataFrame({
'sample_pan': pan_samples,
'is_valid': pan_samples.str.match(pan_regex)
})

display(pan_validation_results)
```
